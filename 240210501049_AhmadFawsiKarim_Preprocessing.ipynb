```python
import re
from Sastrawi.StopWordRemover.StopWordRemoverFactory import StopWordRemoverFactory
from Sastrawi.Stemmer.StemmerFactory import StemmerFactory

print("Library berhasil diimpor.")
```

```python
# Stopword remover
stopword_factory = StopWordRemoverFactory()
stopwords = set(stopword_factory.get_stop_words())

# Stemmer Bahasa Indonesia
stemmer_factory = StemmerFactory()
stemmer = stemmer_factory.create_stemmer()

print("Jumlah stopwords:", len(stopwords))
print("Stopword dan stemmer siap digunakan.")
```

```python
dokumen = [
    {
        "id": "D1",
        "judul": "Pengembangan Laboratorium Komputer Fakultas Teknik",
        "teks": (
            "Fakultas Teknik UNM mengembangkan laboratorium komputer untuk mendukung "
            "kegiatan pembelajaran mahasiswa. Laboratorium dilengkapi komputer baru, "
            "jaringan internet, dan perangkat lunak pembelajaran."
        )
    },
    {
        "id": "D2",
        "judul": "Pelatihan Kecerdasan Buatan bagi Mahasiswa",
        "teks": (
            "Program studi Teknik Komputer menyelenggarakan pelatihan kecerdasan buatan "
            "bagi mahasiswa. Kegiatan membahas dasar pembelajaran mesin, pengolahan data, "
            "dan penerapan kecerdasan buatan dalam bidang pendidikan."
        )
    },
    {
        "id": "D3",
        "judul": "Seminar Teknologi Informasi dan Keamanan Data",
        "teks": (
            "Fakultas Teknik mengadakan seminar teknologi informasi yang membahas keamanan "
            "data. Peserta mendapatkan pengetahuan tentang perlindungan data pribadi, "
            "kata sandi, serta keamanan jaringan komputer."
        )
    },
    {
        "id": "D4",
        "judul": "Pengumuman Perkuliahan Berbasis Daring",
        "teks": (
            "Mahasiswa Fakultas Teknik dapat mengikuti perkuliahan berbasis daring melalui "
            "platform pembelajaran yang telah disediakan. Dosen diharapkan mengunggah materi "
            "dan tugas perkuliahan secara berkala agar kegiatan belajar berjalan dengan baik."
        )
    },
    {
        "id": "D5",
        "judul": "Penerapan Sistem Informasi Akademik",
        "teks": (
            "Universitas meningkatkan layanan akademik melalui penerapan sistem informasi "
            "terintegrasi. Sistem tersebut membantu mahasiswa melakukan pengisian KRS, "
            "melihat jadwal kuliah, memeriksa nilai, dan memperoleh informasi akademik."
        )
    }
]

print("Jumlah dokumen:", len(dokumen))
for d in dokumen:
    print(f'{d["id"]}: {d["judul"]}')
```

```python
def preprocess_text(text):
    # 1. Case folding
    text = text.lower()

    # 2. Cleaning: hanya mempertahankan huruf dan spasi
    text = re.sub(r'[^a-zA-Z\s]', ' ', text)

    # Merapikan spasi
    text = re.sub(r'\s+', ' ', text).strip()

    # 3. Tokenisasi
    tokens = text.split()

    # 4. Stopwords removal
    tokens = [token for token in tokens if token not in stopwords]

    # 5. Stemming
    tokens = [stemmer.stem(token) for token in tokens]

    return tokens

print("Fungsi preprocess_text() berhasil dibuat.")
```

```python
teks_uji = "Mahasiswa Fakultas Teknik sedang mempelajari Teknologi Informasi!"

hasil_uji = preprocess_text(teks_uji)

print("Teks asli   :", teks_uji)
print("Hasil proses:", hasil_uji)
```

```python
for d in dokumen:
    d["tokens_sebelum"] = re.findall(r'[a-zA-Z]+', d["teks"].lower())
    d["tokens_sesudah"] = preprocess_text(d["teks"])

print("Preprocessing 5 dokumen selesai.")
```

```python
for d in dokumen:
    print("=" * 80)
    print(f'{d["id"]} — {d["judul"]}')
    print("\nSebelum preprocessing:")
    print(d["teks"])
    print("\nToken sebelum:")
    print(d["tokens_sebelum"])
    print("\nToken sesudah preprocessing:")
    print(d["tokens_sesudah"])
```

```python
for d in dokumen[:2]:
    print("=" * 80)
    print(f'{d["id"]} — {d["judul"]}')
    print("SEBELUM:")
    print(d["teks"])
    print("\nSESUDAH:")
    print(" ".join(d["tokens_sesudah"]))
    print()
```

```python
for d in dokumen:
    jumlah_sebelum = len(d["tokens_sebelum"])
    jumlah_sesudah = len(d["tokens_sesudah"])
    pengurangan = jumlah_sebelum - jumlah_sesudah
    persentase = (pengurangan / jumlah_sebelum * 100) if jumlah_sebelum > 0 else 0

    d["jumlah_sebelum"] = jumlah_sebelum
    d["jumlah_sesudah"] = jumlah_sesudah
    d["persentase_pengurangan"] = persentase

    print(
        f'{d["id"]}: sebelum={jumlah_sebelum}, '
        f'sesudah={jumlah_sesudah}, '
        f'pengurangan={persentase:.2f}%'
    )
```

Preprocessing teks memberikan dampak positif terhadap kualitas data untuk sistem temu kembali informasi karena teks menjadi lebih seragam dan lebih fokus pada kata yang memiliki makna

biasanya kurang membantu proses pencarian, sedangkan stemming menyamakan berbagai bentuk kata berimbuhan ke bentuk dasarnya. Akibatnya, jumlah token menjadi lebih sedikit
sehingga data lebih ringkas dan dapat mengurangi variasi kata saat proses pencocokan dokumen. Namun, preprocessing perlu dilakukan secara tepat karena penghapusan stopwords atau
stemming yang terlalu agresif dapat menghilangkan informasi yang masih relevan terhadap suatu query.
